Cargar y cruzar datos de Kaggle

```python
import pandas as pd
import os

# Determinar path
current_dir = '/Users/raquelperez/data-engineering-project/Data Science/Determinación de precios Ecommerce/data'

# 1. Cargar las tablas clave para el análisis de precios
items = pd.read_csv(current_dir+'/olist_order_items_dataset.csv')
products = pd.read_csv(current_dir+'/olist_products_dataset.csv')
category_translation = pd.read_csv(current_dir+'/product_category_name_translation.csv')

# 2. Unir productos con sus traducciones de categoría
products = products.merge(category_translation, on='product_category_name', how='left')

# 3. Cruzar la información de ventas con las características del producto
df = items.merge(products, on='product_id', how='inner')

# 4. Vista previa de los datos relevantes para determinación de precios
cols_interes = [
    'product_id', 
    'product_category_name_english', 
    'price', 
    'freight_value', 
    'product_weight_g', 
    'product_photos_qty'
]

print("Forma del dataset fusionado:", df[cols_interes].shape)
print("\nEstadísticas descriptivas de precios:")
print(df[['price', 'freight_value']].describe())
```
